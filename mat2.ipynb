```python
import torch
```

```python
A=torch.tensor([[1.,1.,1.], [1.,1.,1.]])
B=torch.tensor([[0.,1.], [2,3], [4,5]])
```

```python
A.shape
```

```python
B.shape
```

```python
x=A[:,0]
x.requires_grad_(True)  # Same as `x = torch.arange(4.0, requires_grad=True)`
x.grad
```

```python
from numpy import matmul
C = torch.matmul(A,B)
```

```python
y = 3*C[1].T*(x*x)
```

```python
y.backward(torch.ones_like(y))
x.grad
```
