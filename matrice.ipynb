```python
import torch
```

```python
A = torch.tensor([[ 4.,  5.],
                  [-2.,  1.],
                  [ 0.,  6.]])

B = torch.tensor([[ 3.,  0., -1.],
                  [ 2.,  2., 10.]])
```

```python
print(A.shape)
print(B.shape)
```

```python
x = A[:, 1].clone().detach().requires_grad_(True)
```

```python
C = B[0] + torch.matmul(A,B)
```

```python
f = -C[0]*(x*x)
```

```python
f.backward(torch.ones_like(f))
```

```python
print(x.grad)
```
